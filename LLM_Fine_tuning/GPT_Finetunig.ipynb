```python
clinet.files.create(
    file=open("/content/data.jsonl","rb"),
    purpose="fine-tune"
)
```

```python
clinet.files.list()
```

```python

for file in clinet.files.list():
  print(file.id)
  print(100*"=")
  print(file.purpose)
  print(100*"=")
  print(file)
```

```python
clinet.fine_tuning.jobs.create(
  training_file="file-Mz8kuEKjPrVhAsgdeBeP9b",
  model="gpt-4o-2024-08-06",
  suffix="genaibootcamp-live-class",
  method={
    "type": "supervised",
    "supervised": {
      "hyperparameters": {
        "learning_rate_multiplier": 1.0,
        "n_epochs": 3
      }
    }
  }
)
```

```python
clinet.files.list()
```

```python

for job in clinet.fine_tuning.jobs.list():
  print(job.fine_tuned_model)
```
